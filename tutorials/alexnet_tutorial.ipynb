##### Data Setup
The casting data should be in the `casting_data/` folder within this tutorials directory. If you don't have the data yet, you can download it from the dataset sources or use ManufacturingNet's dataset utilities.

```python
from ManufacturingNet import datasets
import os
import zipfile
from pathlib import Path

# Get the current notebook directory
notebook_dir = Path(os.getcwd())
print(f"Notebook directory: {notebook_dir}")

def ensure_casting_data():
    data_dir = notebook_dir / "casting_data"
    if data_dir.exists():
        print(f"✓ Found casting_data at {data_dir}")
        return data_dir

    print(f"\nWARNING: casting_data directory not found at {data_dir}")
    print("\nAttempting to download the CastingData dataset automatically...")

    try:
        datasets.CastingData()
    except Exception as exc:
        print(f"Primary download failed: {exc}")
        print("Retrying with the direct Google Drive fallback...")
        try:
            zip_path = notebook_dir / "CastingData.zip"
            if zip_path.exists():
                zip_path.unlink()
            datasets.download_file_from_google_drive_with_gdown(
                "1qNnLCcq1HlzS0WmOCRlJfNC9ZF26j_6f",
                str(zip_path),
            )
            with zipfile.ZipFile(zip_path, "r") as zf:
                zf.extractall(notebook_dir)
        except Exception as fallback_exc:
            raise RuntimeError(
                f"Automatic dataset download failed: {fallback_exc}"
            ) from fallback_exc

    if not data_dir.exists():
        zip_path = notebook_dir / "CastingData.zip"
        if zip_path.exists() and zipfile.is_zipfile(zip_path):
            with zipfile.ZipFile(zip_path, "r") as zf:
                zf.extractall(notebook_dir)

    if not data_dir.exists():
        raise FileNotFoundError(
            f"Dataset still not found after download. Expected folder: {data_dir}"
        )

    print(f"✓ Downloaded and extracted casting_data to {data_dir}")
    return data_dir


ensure_casting_data()
```

##### Data Validation
Now let's verify the data structure and set up the paths for training.

The Casting dataset is an image dataset with 2 classes. The task that we need to perform using Pretrained Alexnet is classification. ManufacturingNet has also provided different datasets in the package which the user can choose depending on type of application

Pretrained models use Imagefolder dataset from pytorch and image size is (224,224,channels). The pretrained model needs the root folder path of train and test images(Imagefolder format). Manufacturing pretrained models have image resizing feature. 

```python
# Set up paths for the casting dataset
import os
from pathlib import Path

# Define data paths relative to notebook location
notebook_dir = Path(os.getcwd())
train_data_address = str(notebook_dir / "casting_data" / "train")
val_data_address = str(notebook_dir / "casting_data" / "test")

print(f"Train data path: {train_data_address}")
print(f"Validation data path: {val_data_address}")

# Verify data exists
if os.path.exists(train_data_address):
    print(f"✓ Train directory found")
    print(f"  Classes: {os.listdir(train_data_address)}")
else:
    print(f"✗ Train directory not found at {train_data_address}")

if os.path.exists(val_data_address):
    print(f"✓ Validation directory found")
    print(f"  Classes: {os.listdir(val_data_address)}")
else:
    print(f"✗ Validation directory not found at {val_data_address}")
```
